```python
# Chuyển tên cột thành chữ thường và thay khoảng trắng bằng dấu gạch dưới
df.columns = df.columns.str.lower().str.replace(' ', '_')
df = df.rename(columns={'purchase_amount_(usd)': 'purchase_amount'})
```

```python
# Phân chia nhóm tuổi và gán nhãn tương ứng
labels = ['young adult', 'adult', 'middle-aged', 'senior']
df['age_group'] = pd.qcut(df['age'], q=4, labels=labels)
df[['age', 'age_group']].head(10)
```

```python
# Tạo từ điển ánh xạ tần suất mua hàng sang số ngày tương ứng
frequency_mapping = {
    'Weekly': 7,
    'Bi-Weekly': 14,          # Mỗi 2 tuần (tương tự Fortnightly)
    'Fortnightly': 14,        # Mỗi 2 tuần
    'Monthly': 30,           # Hàng tháng
    'Every 3 Months': 90,     # Mỗi 3 tháng (tương tự Quarterly)
    'Quarterly': 90,          # Hàng quý (3 tháng)
    'Annually': 365           # Hàng năm
}

# Áp dụng hàm map để tạo cột mới dạng số
df['purchase_frequency_days'] = df['frequency_of_purchases'].map(frequency_mapping)
```

```python
# Kiểm tra sự trùng lặp giữa hai cột
(df['discount_applied'] == df['promo_code_used']).all()
```

```python
# Loại bỏ cột promo_code_used
df = df.drop('promo_code_used', axis=1)
```

```python
df.to_csv('Shopping Trends And Customer Behaviour Dataset Clean.csv', index=False)
```
